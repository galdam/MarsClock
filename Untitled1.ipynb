```python

def buffer_2_pilimg(width, height, black_ch, color_ch, scale=1):
    palette = {(1,1): (230, 230, 230), # white
            (1, 0): (230, 0, 0),  # Red
             (0, 1): (10, 10, 10), # Black
                }
    a = scale
    img_size = (width*a, height*a)
    img = Image.new( 'RGB', img_size, "green") # Create a new image
    pixels = img.load() # Create the pixel map

    #print(fb.height, 'x', fb.width)
    #print(fb.display_height, 'x', fb.display_width)
    #for y in range(max(fb.height, fb.width)): #range(fb.height):
    #    for x in range(max(fb.height, fb.width)): #range(fb.width):
    
    for y in range(height): #range(fb.height):
        for x in range(width): #range(fb.width):
            i = (y*width)+x
            
            c = palette[(black_ch[i], color_ch[i])]
            for i in range(a):
                for j in range(a):
                    pixels[(x*a)+i,(y*a)+j] = c
    return img
```

```python
def byte_array_to_bit_rev(ch):
    return [byte >> i & 1 for byte in ch for i in reversed(range(8))]

def byte_array_to_bit_fwd(ch):
    return [byte >> i & 1 for byte in ch for i in (range(8))]
```

```python

def get_2b_pixel(buffer, width, x, y):
    """Get the color of a given pixel"""
    pixel_num = (y * width + x)
    index = pixel_num >> 2
    pixel = buffer[index]
    shift = (pixel_num & 0b11) << 1
    return (pixel >> shift) & 0b11
```

```python
def rotation_iterator(rotation, height, width):
    if rotation == 0:
        for y in range(height): #range(fb.height):
            for x in range(width): #range(fb.width):
                yield x, y
    elif rotation == 1:
        raise NotImplementedError()
    elif rotation == 2:
        for y in range(height-1, -1, -1): #range(fb.height):
            for x in range(width-1, -1, -1): #range(fb.width):
                yield x, y
    elif rotation == 3:
        raise NotImplementedError()
    else:
        raise ValueError()
```

```python
def calculate_moon_ellipse(xo, yo, size, phase_pct, fg, bg, outer=True):
        """
        Get the series of ellipses to draw to represent a moon.
        Assumes that the background is white.
        Each phase can be represented as two half ellipses where one is full, and the other is squashed.

        xo, yo - int; x, y origin of where the moon should be drawn
        size - int; size of the moon to be drawn
        phase_pct - float; the percentage of the way through the moon phase
        """
        quart_pct = (phase_pct * 0.25) % 1
        r = size // 2
        x, y = xo+r, yo+r
        # Erase the background
        yield x, y, r, r, bg, True

        # Waxing crescent
        if phase_pct <= 0.25:
            yield x, y, r, r, fg, True, LEFT
            yield x, y, int(r * (1 - quart_pct)), r, fg, True, RIGHT
        # Waxing gibbous
        elif phase_pct <= 0.5:
            yield x, y, r, r, fg, True, LEFT
            yield x, y, int(r * quart_pct), r, bg, True, LEFT
        # Waning gibbous
        elif phase_pct <= 0.75:
            yield x, y, r, r, fg, True, RIGHT
            yield x, y, int(r * (1 - quart_pct)), r, bg, True, RIGHT
        # Waning crescent
        else:
            yield x, y, r, r, fg, True, RIGHT
            yield x, y, int(r * quart_pct), r, fg, True, LEFT
        # Draw the outer ring if enabled or if the moon  is new
        if outer or (phase_pct > 0.90 or phase_pct < 0.1):
            yield x, y, r, r, fg, False
```

```python
def next_suntimes(observer, et, location):
    et_kwargs={'tm_tzone':et.tm_tzone}

    er = ephemerisrelative.EphemerisRelative(
        observer=observer,
        j2k=et.j2kdelta,
        location=location,
        et_kwargs=et_kwargs
    )
    ets = er.rise_elevation_transit_set
    if ets['state'] is not None:
        return None
    if ets['rise'] > et:
        return [('rise', ets['rise']), ('set', ets['set'])]
    
    er = ephemerisrelative.EphemerisRelative(
        observer=observer,
        j2k=et.j2kdelta+1,
        location=location,
        et_kwargs=et_kwargs
    )
    etsn = er.rise_elevation_transit_set
    if etsn['state'] is not None:
        return None
    
    if ets['set'] > et:
        return [('set', ets['set']), ('rise', etsn['rise'])]
    
    return [('rise', etsn['rise']), ('set', etsn['set'])]


next_suntimes('Earth', h.rtc.earth_time, location)
```
