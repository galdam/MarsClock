```python
import numpy as np

def normalize_angle_deg(angle):
    """Normalize angle to [0, 360) degrees."""
    return angle % 360.0

def sun_longitude_from_earth_heliocentric(earth_long_deg, earth_lat_deg=0.0):
    """
    Compute apparent geocentric ecliptic longitude of the Sun
    from Earth's heliocentric ecliptic longitude and latitude.
    (Two-body approximation — no aberration, nutation, etc.)
    """
    # Convert to radians
    lam_earth = np.deg2rad(earth_long_deg)
    beta_earth = np.deg2rad(earth_lat_deg)
    
    # Earth's heliocentric rectangular coordinates
    x_earth = np.cos(beta_earth) * np.cos(lam_earth)
    y_earth = np.cos(beta_earth) * np.sin(lam_earth)
    z_earth = np.sin(beta_earth)
    
    # Sun's geocentric vector is the negative of Earth's
    x_sun, y_sun, z_sun = -x_earth, -y_earth, -z_earth
    
    # Compute Sun's ecliptic longitude and latitude
    lambda_sun = np.arctan2(y_sun, x_sun)
    beta_sun = np.arcsin(z_sun / np.sqrt(x_sun**2 + y_sun**2 + z_sun**2))
    
    # Convert to degrees and normalize
    lambda_sun_deg = normalize_angle_deg(np.rad2deg(lambda_sun))
    beta_sun_deg = np.rad2deg(beta_sun)
    
    return lambda_sun_deg, beta_sun_deg
```
